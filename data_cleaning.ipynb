```python
top_subcategory = (
    store.groupby('Sub-Category')['Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_subcategory
```

```python
top_subcategory.plot(kind='barh')

plt.title("Top 10 Sub-Categories by Sales")
plt.xlabel("Sales")
plt.ylabel("Sub-Category")

plt.show()
```

```python
top_states = (
    store.groupby('State')['Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_states
```

```python
top_states.plot(kind='barh')

plt.title("Top 10 States by Sales")

plt.xlabel("Sales")
plt.ylabel("State")

plt.show()
```

```python
top_city = (
    store.groupby('City')['Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_city
```

```python
segment_sales = (
    store.groupby('Segment')['Sales']
    .sum()
    .sort_values(ascending=False)
)

segment_sales
```
